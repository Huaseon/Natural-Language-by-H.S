```python
import pandas as pd
```

```python
def get_acc(dir: list[str]) -> list[float]:
    if dir:
        df = pd.read_csv(f'{dir[0]}/preds_test.csv')
        return {dir[0]: 1 - (df.iloc[:, 2::3].to_numpy() - df.iloc[:, ::3].to_numpy()).__abs__().mean(),  **get_acc(dir[1:])}
    return {}
```

```python
get_acc(['3-2-10', '3-2-13', '3-2-15', '3-2-20', '3-2-25', 'output/experiment-v1'])
```
